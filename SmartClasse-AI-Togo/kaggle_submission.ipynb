# 🌍 SmartClasse AI Togo — Gemma 4 Good Hackathon

> **Adaptive offline education system for Togolese students powered by Gemma 4 via Ollama**

---

## The Problem

In Togo, **3.2 million children** face critical barriers to quality education:
- 🌐 **No internet** in 70% of rural schools
- 🗣️ **Language barrier** — French-only curricula in a country with 39 local languages
- 👩‍🏫 **1 teacher per 80+ students** on average
- 📉 **Girls dropout rate** reaches 45% before CM2 (final primary year)

## Our Solution: SmartClasse AI

A **fully offline, multi-agent AI education system** that runs locally on a Raspberry Pi or low-cost laptop using **Gemma 4 E4B via Ollama**:

| Agent | Role | Impact |
|-------|------|--------|
| **ADAPTIX** | Adaptive tutor — personalized exercises with cultural context | Each child learns at their own pace |
| **LINGUIX** | Multilingual bridge — translates between French & 5 Togolese languages | No child left behind due to language |
| **DIAGNOSTIX** | Learning gap analyzer | Teacher knows exactly who needs help |
| **PILOTIX** | Teacher dashboard | One teacher manages 80 students effectively |
| **EQUITIX** | Girl dropout risk detector | Early intervention saves school careers |
| **PARENTIX** | Weekly SMS to parents (local language) | Families stay engaged |
| **KULTURIX** | Cultural context retrieval | Math with sorghum, not euros |

## Setup & Installation

````python
# Install dependencies (Kaggle environment)
import subprocess, sys

def install(pkg):
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', pkg])

packages = [
    'fastapi', 'uvicorn[standard]', 'pydantic>=2.5', 'pydantic-settings',
    'ollama', 'sqlalchemy', 'python-dotenv', 'python-multipart',
    'loguru', 'requests', 'cryptography', 'httpx'
]
for pkg in packages:
    install(pkg)

print('✅ Core dependencies installed')
````

````python
# Check if Ollama is available (local run) or use demo mode
import requests

OLLAMA_URL = 'http://localhost:11434'
BACKEND_URL = 'http://localhost:8000'

def check_ollama():
    try:
        r = requests.get(f'{OLLAMA_URL}/api/tags', timeout=3)
        models = [m['name'] for m in r.json().get('models', [])]
        gemma4 = any('gemma4' in m for m in models)
        print(f'✅ Ollama running | Gemma 4 available: {gemma4}')
        print(f'   Available models: {models}')
        return True
    except Exception as e:
        print(f'⚠️  Ollama not running ({e})')
        print('   → Demo mode: showing pre-computed responses')
        return False

LIVE_MODE = check_ollama()
````

**Why Gemma 4 E4B?**
- 4-bit quantized — runs on 4GB RAM (Raspberry Pi 5, cheap laptops)
- Fast inference via Ollama: ~3-5 seconds per response on CPU
- Excellent multilingual capabilities for French + African languages
- Fully offline — no API key, no internet, no cloud

## Demo 2: LINGUIX — Multilingual Translation

A teacher says an instruction in French. LINGUIX translates it live to 5 Togolese languages so every child understands.

````python
DEMO_RESPONSES['linguix_translate'] = {
    'status': 'success',
    'translations': [
        {'language': 'kabyie',  'translated_instruction': 'Cɛ lɛ mɩyaɣ takayɩ nɛ ɩmaɣzɩnɩ 12 nɛ 8.',
         'audio': {'audio_url': '/audio/kabyie_instruction.wav'}},
        {'language': 'ewe',     'translated_instruction': 'Zi ŋutinya ɖe takpla la ŋu eye woŋlɔ 12 kple 8.',
         'audio': {'audio_url': '/audio/ewe_instruction.wav'}},
        {'language': 'haoussa', 'translated_instruction': 'Bude littafin karatunka kuma ƙara 12 da 8.',
         'audio': {'audio_url': '/audio/haoussa_instruction.wav'}},
        {'language': 'mina',    'translated_instruction': 'Yi yayra blɔ ɖe dzo la ŋu eye woŋlɔ 12 kple 8.',
         'audio': {'audio_url': '/audio/mina_instruction.wav'}},
        {'language': 'tem',     'translated_instruction': 'Yaa bukuu ŋ kɔliŋ ŋ yɔ 12 nɛ 8.',
         'audio': {'audio_url': '/audio/tem_instruction.wav'}},
    ]
}

def demo_linguix_translate(instruction, source='french', targets=None):
    if targets is None:
        targets = ['kabyie', 'ewe', 'haoussa', 'mina', 'tem']
    
    if LIVE_MODE:
        try:
            r = requests.post(f'{BACKEND_URL}/agents/linguix/translate_instruction', json={
                'instruction': instruction,
                'source_language': source,
                'target_languages': targets,
                'audio_output': False
            }, timeout=60)
            return r.json()
        except Exception as e:
            print(f'API error: {e} — using demo response')
    
    return DEMO_RESPONSES['linguix_translate']

instruction = 'Ouvrez vos cahiers à la page 12 et calculez 12 plus 8.'
print('=' * 60)
print(f'LINGUIX — Teacher says (French):')
print(f'  "{instruction}"')
print('Translating to 5 Togolese languages...')
print('=' * 60)

result = demo_linguix_translate(instruction)
for t in result.get('translations', []):
    lang = t.get('language', 'unknown')
    text = t.get('translated_instruction', 'N/A')
    audio = t.get('audio', {}).get('audio_url', '')
    flag = {'kabyie': '🇹🇬', 'ewe': '🇹🇬', 'haoussa': '🇳🇬', 'mina': '🇹🇬', 'tem': '🇹🇬'}.get(lang, '🌍')
    print(f"{flag} {lang.upper():10s}: {text}")
    if audio:
        print(f"              🔊 Audio: {audio}")
````
